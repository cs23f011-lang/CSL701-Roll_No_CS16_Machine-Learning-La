```python
#Print only Age and Fare statistics
print("Age Mean :", df["Age"].mean())
print("Age Minimum :", df["Age"].min())
print("Age Maximum :", df["Age"].max())

print("Fare Mean :", df["Fare"].mean())
print("Fare Minimum :", df["Fare"].min())
print("Fare Maximum :", df["Fare"].max())
```

Task 9: Fill Missing Age Values

```python
median_age = df["Age"].median()

df["Age"].fillna(median_age, inplace=True)
```

```python
df["Age"].isnull().sum()
```

Task 10: Filter Surviving Passengers

```python
survivors = df[df["Survived"] == 1]

survivors.head()
```

Task 11: Convert Fare to NumPy Array

```python
fare_array = df["Fare"].to_numpy()

print(fare_array)
```

Task 12: Maximum and Average Fare

```python
print("Maximum Fare =", np.max(fare_array))
print("Average Fare =", np.mean(fare_array))
```

Task 13: Histogram of Age

```python
plt.figure(figsize=(7,5))

plt.hist(df["Age"], bins=20)

plt.title("Histogram of Age")

plt.xlabel("Age")

plt.ylabel("Frequency")

plt.show()
```

Task 14: Bar Chart of Survival Count

```python
sns.countplot(x="Survived", data=df)

plt.title("Survival Count")

plt.show()
```

Task 15: Scatter Plot of Age vs Fare

```python
plt.figure(figsize=(8,6))

plt.scatter(df["Age"], df["Fare"])

plt.xlabel("Age")

plt.ylabel("Fare")

plt.title("Age vs Fare")

plt.show()
```
